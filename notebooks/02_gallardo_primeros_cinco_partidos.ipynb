# 02. Gallardo: los primeros cinco partidos

## Mi pregunta

Que ha hecho Marcelo Gallardo en sus primeros cinco partidos con Ecuador, y que se puede decir con solo cinco partidos, sin exagerar.

## Por que importa

Gallardo llego en septiembre de 2026, con contrato hasta el Mundial de 2030, y reemplazo a Sebastian Beccacece. Cinco partidos no definen a un entrenador, pero si marcan una primera tendencia. Quiero medirla con los numeros que tengo, y decir en voz alta lo que no tengo.

Enner Valencia se retiro de la seleccion el 1 de julio de 2026, con 109 partidos y 49 goles, como maximo goleador. Ese dato cambia el contexto. Ecuador ya no tiene a su goleador historico, y tiene que construir el gol con otros jugadores.

```python
import pathlib
import pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["axes.titlesize"] = 13
plt.rcParams["axes.labelsize"] = 11
BASE = pathlib.Path.cwd().parent if pathlib.Path.cwd().name == "notebooks" else pathlib.Path.cwd()
DATA = BASE / "data"
CHARTS = BASE / "charts"
CHARTS.mkdir(exist_ok=True)
print("Base:", BASE)
print("Charts:", CHARTS)

gal = pd.read_csv(DATA / "gallardo_partidos_2026.csv")
print(gal.to_string(index=False))
print()
print("Balance verificado en 2026: 2 victorias, 2 empates, 1 derrota")
print("Goles a favor: 5. Goles en contra: 4. Diferencia: +1")
print("Puntos si fueran partidos de liga, 3 por victoria: 8 de 15")
```

```python
con_marcador = gal.dropna(subset=["goles_a_favor"]).copy()
con_marcador["goles_a_favor"] = con_marcador["goles_a_favor"].astype(int)
con_marcador["goles_en_contra"] = con_marcador["goles_en_contra"].astype(int)
print(con_marcador[["rival", "resultado", "fecha", "goles_a_favor", "goles_en_contra", "sede"]].to_string(index=False))
print()
print("Partidos con marcador verificado:", len(con_marcador), "de 5")
print("En esos 3 partidos: 2 victorias, 1 derrota, 4 goles a favor, 3 en contra")
print("Los dos empates, ante Japon y Panama, suman juntos 1 gol a favor y 1 en contra,")
print("porque el total es 5 a 4. El reparto entre Japon y Panama no esta verificado,")
print("asi que no lo asigno a un partido concreto.")
```

```python
plt.figure()
x = con_marcador["rival"]
plt.bar(x, con_marcador["goles_a_favor"], label="Goles a favor", color="#1f4e79")
plt.bar(x, con_marcador["goles_en_contra"], label="Goles en contra", color="#c0392b", alpha=0.75)
plt.ylabel("Goles")
plt.title("Gallardo 2026: marcador verificado por partido (3 de 5)")
plt.legend()
plt.tight_layout()
plt.savefig(CHARTS / "03_gallardo_marcadores.png", dpi=150)
plt.show()
```

```python
plt.figure()
categorias = ["Victorias", "Empates", "Derrotas"]
valores = [2, 2, 1]
plt.bar(categorias, valores, color=["#1f4e79", "#7f8c8d", "#c0392b"])
plt.ylabel("Partidos")
plt.title("Gallardo, primeros 5 partidos en 2026: 2-2-1, 5 goles a favor, 4 en contra")
for i, v in enumerate(valores):
    plt.text(i, v + 0.05, str(v), ha="center")
plt.ylim(0, 2.6)
plt.tight_layout()
plt.savefig(CHARTS / "04_gallardo_balance.png", dpi=150)
plt.show()
```

## Lo que encontre

En cinco partidos, Gallardo suma 2 victorias, 2 empates y 1 derrota, con 5 goles a favor y 4 en contra. La derrota fue 0-3 ante Corea del Sur. Las victorias fueron 2-0 ante Costa Rica el 13 de noviembre en Quito y 2-0 ante Bolivia el 17 de noviembre en Guayaquil. Los empates fueron ante Japon y Panama.

Los dos partidos de noviembre se ganaron sin encajar. Eso es un dato positivo, pero es una muestra de dos partidos, no una tendencia defensiva probada. El sistema proyectado en noviembre era el 1-4-2-3-1, pero Global Sports Archive registro un 4-2-2-2 ante Costa Rica. Por eso trato el 4-2-3-1 como proyectado, no como confirmado sobre el campo.

Kendry Paez si estaba en la convocatoria de 29. Estuvo en el banquillo ante Costa Rica, sin minutos. Tiene 28 partidos y 2 goles con la seleccion a 17 de noviembre de 2026, y juega en el Chelsea Sub-21. Leonardo Campana fue omitido. Tiene 20 partidos y 1 gol, su ultima convocatoria fue el 27 de marzo de 2026 por lesion, y su club no esta confirmado, porque una fuente lo lista en New England Revolution y una nota anterior decia que no tenia club.

## Lo que significa

Gallardo empezo con un balance positivo y con dos porterias a cero seguidas en noviembre, pero con una derrota dura por 0-3 en el camino. Con cinco partidos no puedo hablar de un estilo consolidado. Lo que si puedo decir es que el grupo respondio en Quito y en Guayaquil, que Moises Caicedo marco en los dos partidos, y que el sistema de noviembre todavia esta por confirmar sobre el campo.
