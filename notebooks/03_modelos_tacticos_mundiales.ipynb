# 03. Modelos tacticos mundiales

## Mi pregunta

Que hicieron de verdad los grandes modelos tacticos recientes, con numeros de liga y de torneo, antes de copiar nada para Ecuador.

## Por que importa

Copiar un nombre famoso sin mirar sus numeros es el error mas facil en el analisis tactico. El Liverpool de Klopp, el PSG de Luis Enrique, el Leverkusen de Xabi Alonso, el Barcelona de Xavi, el Chelsea de Maresca, la Alemania de Nagelsmann y los Paises Bajos de Koeman no jugaban igual, no tenian los mismos jugadores y no lograron lo mismo. Quiero poner sus numeros lado a lado, con la temporada y la fuente.

```python
import pathlib
import pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["axes.titlesize"] = 13
plt.rcParams["axes.labelsize"] = 11
BASE = pathlib.Path.cwd().parent if pathlib.Path.cwd().name == "notebooks" else pathlib.Path.cwd()
DATA = BASE / "data"
CHARTS = BASE / "charts"
CHARTS.mkdir(exist_ok=True)
print("Base:", BASE)
print("Charts:", CHARTS)

mod = pd.read_csv(DATA / "modelos_tacticos.csv")
print(mod[["entrenador", "equipo", "temporada", "sistema_base", "puntos_liga", "goles_a_favor", "goles_en_contra", "posesion_porcentaje"]].to_string(index=False))
```

```python
liga = mod.dropna(subset=["puntos_liga"]).copy()
liga["diferencia"] = liga["goles_a_favor"] - liga["goles_en_contra"]
print(liga[["entrenador", "equipo", "temporada", "sistema_base", "puntos_liga", "victorias", "empates", "derrotas", "goles_a_favor", "goles_en_contra", "diferencia", "posesion_porcentaje", "xg", "xga"]].to_string(index=False))
print()
print("Mejor diferencia de goles en liga: Leverkusen 2023-24, +65, con 90 puntos y 0 derrotas")
print("Mejor ataque en liga: PSG 2024-25, 92 goles")
print("Mejor posesion en liga: PSG 2024-25, 68.4 por ciento")
print("Menor xGA en liga: Leverkusen 2023-24, 30.46, seguido por PSG 2024-25, 31.68")
```

```python
plt.figure()
plt.barh(liga["entrenador"] + " (" + liga["temporada"] + ")", liga["puntos_liga"], color="#1f4e79")
plt.xlabel("Puntos en liga")
plt.title("Modelos tacticos: puntos en liga en la temporada de referencia")
plt.tight_layout()
plt.savefig(CHARTS / "05_modelos_puntos_liga.png", dpi=150)
plt.show()
```

```python
plt.figure()
x = range(len(liga))
plt.bar([i - 0.2 for i in x], liga["goles_a_favor"], width=0.4, label="Goles a favor", color="#1f4e79")
plt.bar([i + 0.2 for i in x], liga["goles_en_contra"], width=0.4, label="Goles en contra", color="#c0392b")
plt.xticks(list(x), liga["entrenador"], rotation=20, ha="right")
plt.ylabel("Goles en liga")
plt.title("Modelos tacticos: ataque y defensa en liga")
plt.legend()
plt.tight_layout()
plt.savefig(CHARTS / "06_modelos_goles_liga.png", dpi=150)
plt.show()
```

```python
plt.figure()
liga2 = liga.dropna(subset=["posesion_porcentaje", "xg"])
plt.scatter(liga2["posesion_porcentaje"], liga2["xg"], s=120, color="#1f4e79")
for _, r in liga2.iterrows():
    plt.annotate(r["entrenador"], (r["posesion_porcentaje"], r["xg"]), fontsize=9)
plt.xlabel("Posesion en liga, porcentaje")
plt.ylabel("xG en liga")
plt.title("Posesion y xG: tener el balon no garantiza crear mas")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(CHARTS / "07_modelos_posesion_xg.png", dpi=150)
plt.show()
print("Aviso sobre el PPDA: las fuentes dan valores entre 8.05 y 9.3 para el Liverpool de Klopp.")
print("No presento un unico PPDA definitivo, porque no es comparable entre fuentes.")
print("Aviso sobre jugadas a balon parado: ningun porcentaje esta verificado para estos modelos.")
```
