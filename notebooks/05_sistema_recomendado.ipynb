La media de ajuste por lineas es 4.29 sobre 5, la mas alta de los tres sistemas, y el sistema proyectado por Gallardo en noviembre tambien era el 4-2-3-1, aunque Global Sports Archive registro un 4-2-2-2 ante Costa Rica y por eso no lo doy por confirmado.

Kevin marco 10 goles en 31 partidos de liga con Union Saint-Gilloise en 2025-26. Kevin es el cambio natural cuando el partido pide otro delantero con mas presencia.

## Forma de construir

En posesion, el equipo pasa a un 3-2-5. Un lateral sube, el otro se queda, un pivote baja entre los centrales y el mediapunta se mete entre lineas. Pacho, con un 95 por ciento de pase en la Ligue 1 en 2025-26 segun FootballCritic, es el central que saca el balon. Caicedo se ofrece siempre como primer pase. Vite recibe entre lineas y Plata y Sarmiento abren el campo. La salida no es larga por sistema. Es corta, con Galindez como apoyo, y larga solo si el rival presiona con dos delanteros y deja espacio a la espalda.

## Presion y bloque defensivo

La presion se activa en tres momentos. Primero, cuando el rival juega hacia su lateral y le queda el balon en el pie debil. Segundo, cuando el central rival recibe de espaldas y no tiene pase claro al pivote. Tercero, cuando el rival pierde el balon en su campo y Caicedo puede ir a robar en 3 segundos. No presiono todo el partido. Con 5 partidos de Gallardo y sin datos de eventos completos, no puedo prometer una presion alta sostenida, y no voy a inventar un PPDA para Ecuador.

Sin balon, el bloque es medio, en 4-4-2, con Vite al lado de Jordy Caicedo. La primera linea no persigue al portero. Cierra el pase al pivote rival y empuja el juego a la banda. Caicedo y Franco no salen los dos a la vez. Uno presiona y el otro cubre. Esa regla protege a Torres y a Pacho, que no tienen a Ordonez detras en noviembre.

## Transiciones

Cuando Ecuador roba, el primer pase es a Vite o a Plata, no un balon largo sin destino. Plata tiene 8 asistencias en total con Flamengo en 2025-26 y Sarmiento tiene velocidad para correr al espacio. Jordy Caicedo ataca el primer palo y el extremo contrario llega al segundo. Cuando Ecuador pierde, Caicedo hace la falta tactica si el rival sale con ventaja, y Franco se queda para cubrir la zona central. No hay transicion sin un pivote en su sitio.

## Jugadas a balon parado

No tengo porcentajes verificados de jugadas a balon parado para ningun modelo, y no voy a inventarlos. Lo que si puedo decir con los datos es quien remata. Torres tiene 5 goles con la seleccion en 53 partidos, Pacho tiene 2 en 43, Hincapie tiene 3 en 58 y Estupinan tiene 5 en 60. En los corners a favor, Torres y Pacho van al area, y Caicedo se queda en la frontal para el rebote. En los corners en contra, Jordy Caicedo defiende el primer palo y Galindez, con 11 porterias a cero en Huracan en 2026, manda en el area.
