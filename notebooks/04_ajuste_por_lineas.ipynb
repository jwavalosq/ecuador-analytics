# 04. Ajuste por lineas

## Mi pregunta

Que sistema encaja con cada linea de Ecuador, y donde esta el riesgo, antes de elegir un sistema para todo el equipo.

## Por que importa

Un sistema puede ser perfecto para el mediocampo y malo para los centrales, o bueno para los extremos y malo para el delantero. Si solo miro el total, escondo el problema. Por eso puntuo cada sistema por grupo de posiciones, del 1 al 5, con la razon escrita al lado de cada nota. La escala es mi juicio analitico, basado en los datos verificados de los cuadernos 01 a 03, no un dato medido en el campo.

```python
import pathlib
import pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["axes.titlesize"] = 13
plt.rcParams["axes.labelsize"] = 11
BASE = pathlib.Path.cwd().parent if pathlib.Path.cwd().name == "notebooks" else pathlib.Path.cwd()
DATA = BASE / "data"
CHARTS = BASE / "charts"
CHARTS.mkdir(exist_ok=True)
print("Base:", BASE)
print("Charts:", CHARTS)

ajuste = pd.DataFrame({
    "grupo": ["Portero", "Centrales", "Laterales", "Doble pivote / mediocentro", "Mediapunta", "Extremos", "Delantero centro"],
    "4-2-3-1 (Maresca, Nagelsmann, Koeman)": [4, 4, 4, 5, 4, 5, 4],
    "4-3-3 (Klopp, Luis Enrique, Xavi)": [3, 3, 4, 3, 3, 4, 3],
    "3-4-2-1 (Alonso)": [3, 3, 5, 3, 5, 3, 4],
})
print(ajuste.to_string(index=False))
print()
for col in ajuste.columns[1:4]:
    print(col, "media:", round(ajuste[col].mean(), 2))
print()
print("El 4-2-3-1 lidera en 5 de los 7 grupos y empata o gana en el resto salvo en laterales,")
print("donde el 3-4-2-1 puntua 5 por el uso de carrileros, y en mediapunta, donde tambien puntua 5.")
```

```python
medias = pd.Series({"4-2-3-1": 4.29, "4-3-3": 3.29, "3-4-2-1": 3.71})
plt.figure()
plt.bar(medias.index, medias.values, color=["#1f4e79", "#7f8c8d", "#2e7d6b"])
plt.ylabel("Puntuacion media de ajuste, 1 a 5")
plt.title("Ajuste por lineas: media por sistema, con 7 grupos de posiciones")
plt.ylim(0, 5)
for i, v in enumerate(medias.values):
    plt.text(i, v + 0.1, f"{v:.2f}", ha="center")
plt.tight_layout()
plt.savefig(CHARTS / "08_ajuste_por_sistema.png", dpi=150)
plt.show()
```

```python
ajuste = pd.DataFrame({
    "grupo": ["Portero", "Centrales", "Laterales", "Doble pivote", "Mediapunta", "Extremos", "Delantero"],
    "4-2-3-1": [4, 4, 4, 5, 4, 5, 4],
    "4-3-3": [3, 3, 4, 3, 3, 4, 3],
    "3-4-2-1": [3, 3, 5, 3, 5, 3, 4],
})
plt.figure(figsize=(11, 6))
x = range(len(ajuste))
ancho = 0.25
plt.bar([i - ancho for i in x], ajuste.iloc[:, 1], width=ancho, label="4-2-3-1", color="#1f4e79")
plt.bar(x, ajuste.iloc[:, 2], width=ancho, label="4-3-3", color="#7f8c8d")
plt.bar([i + ancho for i in x], ajuste.iloc[:, 3], width=ancho, label="3-4-2-1", color="#2e7d6b")
plt.xticks(list(x), ajuste["grupo"], rotation=18, ha="right")
plt.ylabel("Puntuacion de ajuste, 1 a 5")
plt.title("Ajuste por lineas: cada sistema, grupo por grupo")
plt.legend()
plt.ylim(0, 5.6)
plt.tight_layout()
plt.savefig(CHARTS / "09_ajuste_por_grupo.png", dpi=150)
plt.show()
```

## Lo que encontre, grupo por grupo

Portero, 4-2-3-1 con 4, 4-3-3 con 3, 3-4-2-1 con 3. Galindez tiene 39 anos y esta en forma, con un 74.0 por ciento de paradas en Huracan en 2026 segun Flashscore. Un doble pivote reduce los tiros que le llegan y le pide menos salidas largas. Un 4-3-3 con un solo mediocentro lo expone mas.

Centrales, 4-2-3-1 con 4, 4-3-3 con 3, 3-4-2-1 con 3. Pacho tiene minutos y nivel de PSG, Torres tiene minutos en Internacional, pero Hincapie solo tiene 67 minutos en la Premier League en 2026-27 y Ordonez tiene 0 minutos por lesion. Una linea de 3 centrales suena protectora, pero pide un tercer central en forma, y Ecuador no lo tiene en noviembre. Con 2 centrales y un doble pivote delante, la proteccion llega del mediocampo, no de un tercer central sin minutos.

Laterales, 4-2-3-1 con 4, 4-3-3 con 4, 3-4-2-1 con 5. Estupinan y Preciado pueden subir y atacar. El 3-4-2-1 les da el carril entero, y por eso puntua mas aqui. Pero ese punto extra cuesta en el centro, donde Ecuador es mas fuerte.

Doble pivote y mediocentro, 4-2-3-1 con 5, 4-3-3 con 3, 3-4-2-1 con 3. Este es el grupo que decide. Caicedo tiene 67 partidos y 3 goles, con 33 partidos y 32 titularidades en la Premier League en 2025-26, y Franco tiene 64 partidos y 50 partidos en total con Atletico Mineiro en 2026. Juntos forman un doble pivote con experiencia, con minutos y con nivel de club alto. Un 4-3-3 con un solo pivote desperdicia a Franco y deja a Caicedo solo ante la transicion.

Mediapunta, 4-2-3-1 con 4, 4-3-3 con 3, 3-4-2-1 con 5. Vite tiene 26 partidos y 1 gol, con 38 partidos y 2.897 minutos en Pumas en 2025-26. Paez tiene 28 partidos y 2 goles, con solo 5 partidos y 229 minutos en el Chelsea Sub-21 en 2026-27. El 3-4-2-1 usa dos mediapuntas y aprovecha a los dos a la vez. El 4-2-3-1 usa uno, con el otro como cambio. Esa es una ventaja real del 3-4-2-1, y la reconozco.

Extremos, 4-2-3-1 con 5, 4-3-3 con 4, 3-4-2-1 con 3. Plata tiene 55 partidos y 9 goles, con 52 partidos, 6 goles y 8 asistencias en total con Flamengo en 2025-26. Sarmiento tiene 27 partidos y 2 goles, Mercado tiene 12 partidos y 1 gol, con 7 goles y 4 asistencias en liga con Sparta en 2025-26, y Nilson Angulo tiene 22 partidos y 3 goles. El 4-2-3-1 les da banda y libertad, con un mediapunta que les conecta. El 3-4-2-1 no tiene extremos puros, y eso desperdicia a Plata.

Delantero centro, 4-2-3-1 con 4, 4-3-3 con 3, 3-4-2-1 con 4. Jordy Caicedo marco 14 goles en 29 partidos de liga con Huracan en 2026. Kevin Rodriguez marco 10 goles en 31 partidos de liga con Union Saint-Gilloise en 2025-26. Los dos necesitan un mediapunta cerca y extremos que centren. El 4-2-3-1 y el 3-4-2-1 les dan ese apoyo. El 4-3-3 de Klopp pide un delantero que presione sin parar, y eso no es lo que mejor hace Jordy Caicedo.

## Lo que significa

La media de ajuste es 4.29 para el 4-2-3-1, 3.71 para el 3-4-2-1 y 3.29 para el 4-3-3. El 4-2-3-1 gana porque usa el mejor grupo de Ecuador, el doble pivote Caicedo y Franco, protege a los centrales, deja a Plata y a Sarmiento en banda, y pone a Vite o a Paez detras de un delantero en forma. El 3-4-2-1 es la alternativa real, no el 4-3-3, porque usa dos mediapuntas y dos carrileros, pero pide un tercer central que hoy no esta en forma.
