Como se juega el plan principal en un partido real, que se hace cuando el marcador cambia, y que alternativa tengo si el plan principal no funciona.

## Por que importa

Un plan sin alternativa es una apuesta. Si Ecuador va perdiendo y no tiene un cambio de sistema preparado, el entrenador improvisa. Si va ganando y no sabe como cerrar, regala el partido. Aqui dejo el plan de partido, los cambios por escenario y la alternativa completa, con sus riesgos escritos.

## Plan de partido, 4-2-3-1

Los primeros 15 minutos son para salir con el balon, sin regalar corners y para medir la presion rival. De 15 a 30 minutos, Estupinan sube y el equipo busca a Plata en el uno contra uno. De 30 a 45, Vite recibe entre lineas y Jordy Caicedo ataca el primer palo en los centros. Al descanso, el ajuste es de presion, no de sistema, salvo lesion.

De 45 a 60 minutos, si falta profundidad, entra Mercado o Nilson Angulo por Sarmiento. De 60 a 75, si falta gol, entra Paez por Vite. Si falta presencia en el area, entra Kevin Rodriguez por Jordy Caicedo. De 75 a 90, el partido se cierra segun el marcador.

## Cuando Ecuador va perdiendo

Si Ecuador va perdiendo por un gol despues del minuto 60, paso al 3-4-2-1. Hincapie entra por un lateral o por Torres, segun quien este mas cansado, y el equipo juega con tres centrales, dos carrileros y dos mediapuntas. Plata pasa a mediapunta derecho y Paez entra como mediapunta izquierdo. Jordy Caicedo se queda solo arriba, con Kevin Rodriguez como segundo delantero si el partido pide dos puntas en los ultimos 15 minutos. El riesgo es claro. Con tres centrales y dos carrileros, el equipo pierde un extremo puro y queda expuesto a la contra si Caicedo y Franco suben los dos. Por eso uno de los dos se queda siempre.

## Cuando Ecuador va ganando

Si Ecuador va ganando por un gol despues del minuto 75, paso a un 4-4-2 sin balon. Vite baja a la banda, Plata baja a la otra banda, y Jordy Caicedo se queda con Kevin Rodriguez o con Mercado arriba para correr a la contra. Entra Alcivar o Castillo por Vite o por Plata, segun quien este mas cansado, para tener un tercer mediocentro. El equipo no se mete en su area. Defiende en bloque medio, en la linea del centro del campo, y sale con dos pases a la banda cuando roba. Si va ganando por dos goles, el cambio es el mismo, pero con menos prisa por salir. La posesion se cuida en campo propio, con Galindez como apoyo.

## Cuando el partido esta empatado

Si el partido esta empatado despues del minuto 70, no cambio el sistema. Cambio a los jugadores. Paez entra por Vite para tener mas gol desde fuera del area, y Mercado entra por Sarmiento para tener mas centros. El doble pivote se mantiene. Un empate no se rompe regalando el centro del campo.

## Alternativa completa: 3-4-2-1, en la linea de Xabi Alonso

La alternativa es el 3-4-2-1 del Leverkusen de Alonso en 2023-24, que gano la Bundesliga con 90 puntos y 0 derrotas, con 89 goles a favor y 24 en contra. En Ecuador, el once seria Galindez, Torres, Pacho y Hincapie como tres centrales, Preciado y Estupinan como carrileros, Caicedo y Franco en el centro, Plata y Paez como mediapuntas, y Jordy Caicedo arriba.

Esta alternativa tiene dos ventajas reales. Primero, pone a Paez y a Plata juntos detras del delantero, y los dos tienen gol. Plata tiene 9 goles con la seleccion y Paez tiene 2. Segundo, da el carril entero a Estupinan y a Preciado, que son laterales de ataque.

Tambien tiene dos riesgos reales. Primero, pide tres centrales en forma, y en noviembre Hincapie solo tiene 67 minutos en la Premier League en 2026-27 y Ordonez tiene 0 minutos por lesion. Segundo, no tiene extremos puros, y eso deja fuera a Sarmiento, a Mercado y a Nilson Angulo como titulares. Por eso es alternativa, no plan principal. Se usa cuando el equipo va perdiendo y necesita dos mediapuntas, o cuando el rival juega con dos delanteros y Ecuador necesita un tercer central.

## Lo que no prometo

No prometo que el 4-2-3-1 vaya a ganar el Mundial de 2030, ni que el 3-4-2-1 vaya a funcionar contra cualquier rival. Con cinco partidos de Gallardo, sin datos de eventos completos y con las temporadas de club mezcladas entre fuentes, lo que tengo es una recomendacion justificada, no una prediccion. El proximo paso es medir el plan en los proximos partidos, con minutos, con pases y con presiones, y cambiar el once cuando los datos cambien.
